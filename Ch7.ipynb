```python
# Scope and Namespaces in Python
# Local scope
def show_name():
    name = "Waqas"
    print(name)
show_name()
```

```python
# Global scope
course = "Python"

def show_course():
    print(course)

show_course()
print(course)
```

```python
name = "Waqas"

def greet():
    print("Hello",name)
    
greet()
print(name)
```

```python
name = "Waqas"

def student():
    name = "Ali"
    print(name)
    
student()
print(name)
```

```python
# Variable shadowing
name = "Global variable"

def student():
    name = "Local variable"
    print(name)
    
student()
print(name)
```

```python
# global keyword
count = 0

def increase_count():
    global count
    count += 1
    
increase_count()
print(count)
```

```python
def increase_count(count):
    return count + 1

count = 0
count = increase_count(count)
print(count)
```

```python
# Enclosing scope
def outer():
    message = "Hello"
    
    def inner():
        print(message)
    
    inner()
outer()
```

```python
# nonlocal keyword
def outer():
    count = 0
    
    def inner():
        nonlocal count
        count += 1
        print(count)
    inner()
outer()
```

```python
# Built in scope
numbers = [1,2,3,4]
print(len(numbers))
```

```python
# LEGB Rule
x = "global"
def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)
    inner()
outer()
```

```python
# Namespace concept
x = 100
def show():
    x = 50
    print(x)

show()
print(x)
```

```python
# locals() and globals()
course = "Python"
def show():
    name = "Waqas"
    print(locals())
show()
```

```python
from collections import deque
numbers = deque([10,20,30])

numbers.append(40)
numbers.appendleft(5)

print(numbers)
```

```python
from collections import deque

numbers = deque([10, 20, 30, 40])

numbers.pop()        # Remove from the right
numbers.popleft()    # Remove from the left

print(numbers)
```
