```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print("Money deposited.",amount)
        
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance.")
        else:
            self.balance -= amount
            print("Money withdrawn.",amount)
    def show_balance(self):
        print("Balance:", self.balance)
        
account1 = BankAccount("Waqas", 10000)
account1.deposit(5000)
account1.show_balance()
```

```python
# Encapsulation using @property
class student:
    def __init__(self, marks):
        self.marks = marks
    
    @property
    def marks(self):
        return self._marks
    
    @marks.setter
    def marks(self, value):
        if value < 0:
            raise ValueError("Marks cannot be negative.")
        self._marks = value
        
Student1 = student(85)
Student1.marks = 90
print(Student1.marks)
    
```

```python
# Hybrid Inheritance
class Person:
    def show_person(self):
        print("Person")
    
class Student(Person):
    def show_student(self):
        print("Student")
    
class Employee(Person):
    def show_employee(self):
        print("Employee")

class Teaching_Assistant(Student, Employee):
    def show_ta(self):
        print("Teaching Assistant")
        
ta = Teaching_Assistant()
ta.show_person()
ta.show_student()
ta.show_employee()
ta.show_ta()
```

```python
# Abstraction
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
    
class Upi_Payment(Payment):
    def pay(self, amount):
        print("Paid", amount, "Using Upi")
        
payment = Upi_Payment()

payment.pay(50)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
    
class Dog(Animal):
    def sound(self):
        print("Barking")
        
class Cat(Animal):
    def sound(self):
        print("meowing")
    
dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass

class Email(Notification):
    def send(self, message):
        print("Email: ", message)

class SMS(Notification):
    def send(self, message):
        print("SMS: ", message)
    
email = Email()
sms = SMS()

email.send("Hello Waqas!")
sms.send("Hi, Waqas")
```

```python
# Composition
class Engine:
    def start(self):
        print("Engine started")
    
class Car:
    def __init__(self):
        self.engine = Engine()
        
    def start_car(self):
        self.engine.start()
        print("Car is ready")

car = Car()
car.start_car()
```
