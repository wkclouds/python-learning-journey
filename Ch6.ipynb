```python
# Returning Multiple Values
def calculate(a, b):
    total = a + b
    difference = a - b
    return total, difference

sum_result, diff_result = calculate(20,10)
print(sum_result)
print(diff_result)
```

```python
def calculate_marks(maths, python):
    total = maths + python
    average = total / 2
    return total, average

total_result, average_result = calculate_marks(80, 90)
print(total_result)
print(average_result)
```

```python
def add_item(item, items = None):
    if items is None:
        items = []
        
    items.append(item)
    return items

print(add_item("A"))
print(add_item("B"))
```

```python
# Keyword Argument
def student_info(name, age, course):
    print("Name",name)
    print("Age",age)
    print("Course",course)
    
student_info(age = 21, course = "Python", name = "Waqas")
```

```python
# *args (used when don't know how many arguments are passed)
def add_numbers(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(add_numbers(90,20,10))
print(add_numbers(3,4,5,6,7,8))
```

```python
# **kwargs
def show_profile(**details):
    for key, value in details.items():
        print(key, value)
        
show_profile(name = "Waqas", Age = 21, Course = "Python")
```

```python
# Function parameter order
def student_report(name, age = 21, *marks, **details):
    print("Name : ",name)
    print("Age : ", age)
    print("Marks : ",marks)
    print("Details : ",details)
    
student_report("Waqas",21,47,39, course = "Python", city = "Mingora Swat")
```

```python
# Lambda Function --> small anonymous function
# this is normal function
def square(number):
    return number * number

print(square(5))
```

```python
square = lambda number: number * number
print(square(5))
```

```python
add = lambda a,b: a + b
print(add(89, 37))
```

```python
# map() --> applies function to every item of an iterable
numbers = [1,2,3,4]
squares = map(lambda number: number, numbers)
print(list(squares))
```

```python
names = ["Waqas","Afnan","Ihsan"]
upper_names = map(str.upper, names)
print(list(upper_names))
```

```python
# filter() --> select items from an iterable based on a condition
numbers = [1,2,3,4,5,6]
even_numbers = filter(lambda number: number % 2 == 0, numbers)
print(list(even_numbers))
```

```python
# reduce() --> combines all items of an iterable in to a final value
from functools import reduce
numbers = [1,2,3,4]
total = reduce(lambda a, b: a + b, numbers)
print(total)
```

```python
# Recursion --> Function calling itself
def factorial(number):
    if number == 1:
        return 1
    return number * factorial (number-1)
print(factorial(5))
```

```python
# Nested Functions
def outer():
    print("Outer function started")
    
    def inner():
        print("Inner function started")
        
    inner()
outer()
```

```python
# Closures
def multiplier(number):
    def multiply(value):
        return value * number
    return multiply

double = multiplier(2)
print(double(5))
```

```python
def make_discount(discount):
    def calculate(price):
        return price - (price * discount / 100)
    return calculate

ten_percent_off = make_discount(10)
twenty_percent_off = make_discount(20)

print(ten_percent_off(100))
print(twenty_percent_off(100))
```

```python
def make_greeting(name):
    def greet():
        print("Hello", name)
    return greet

waqas_greeting = make_greeting("Waqas")
ali_greeting = make_greeting("Ali")

waqas_greeting()
ali_greeting()
```

```python
def square(number: int) ->int:
    return number * number

print(square(5))
```

```python
# Higher order functions
def double(n):
    return n * 2

def run_func(func,  number):
    return func(number)

print(run_func(double, 5))
```
