```python
n = 4

# Upper half
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")

    for j in range(2 * i - 1):
        print("*", end="")

    print()

# Lower half
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")

    for j in range(2 * i - 1):
        print("*", end="")

    print()
```

```python
# zip()
names = ["Waqas", "Ali", "Ahmed"]
marks = [90, 85, 95]
for name, mark in zip(names, marks):
    print(name, mark)
```

```python
names = ["Waqas", "Ali", "Ahmed"]
marks = [90, 85, 67]
subjects = ["Python", "Excel", "Statistics"]
for name, mark, subject in zip(names, marks, subjects):
    print(name, mark, subject)
```

```python
# zip() + enumerat()
names = ["Waqas", "Ali", "Ahmed"]
marks = [90, 85, 67]
for number, (name, mark) in enumerate(zip(names, marks), start = 1):
    print(number, name, mark)
```

```python
students = ["Ali", "Ahmed", "Waqas"]
scores = [85, 72, 91]

for student, score in zip(students, scores):
    if score >= 80:
        print(student, "Passed with good marks")
```

```python
names = ["Waqas", "Ali", "Ahmed"]
iterator = iter(names)

while True:
    try:
        name = next(iterator)
        print(name)
    except StopIteration:
        break
```
